# AI Production Studio — Studio de voix dans Colab

**Tout se passe ici. Aucun logiciel Windows à installer.**

1. Sélectionnez votre GPU L4 dans **Exécution → Modifier le type d’exécution**.
2. Cliquez sur **Tout exécuter**, puis connectez Google Drive quand Colab le demande. La première installation peut prendre plusieurs minutes.
3. En bas, choisissez le profil et la variante, puis cliquez sur **Créer et écouter**.

Les lecteurs audio apparaissent pour écouter vos essais. Après votre clic sur **Enregistrer cette voix**, les WAV, JSON et références réutilisables sont conservés dans **Mon Drive / AI Production Studio / Voix**. Aucun téléchargement manuel ni ZIP à envoyer à cette conversation n’est nécessaire pour les conserver. Le catalogue du dossier est mis à jour. Vous choisissez vous-même les voix à conserver après écoute.

16 profils cibles : timbres masculins/féminins, petite enfance, enfant, adolescent, environ 20/30/40/50 ans et senior. 1 à 99 variantes par profil, créées seulement quand vous lancez le bouton ; aucune boucle sur tous les profils. Pour vos deux variantes par profil, utilisez 1 puis 2. Le même profil/numéro retrouve sa référence sauvegardée. Âges et stabilité ne sont pas garantis ; vérifier chaque résultat et la diction arabe.

Méthode : Qwen VoiceDesign crée une référence fictive en français ; Chatterbox la réutilise en FR/AR. Qwen ne prend pas directement en charge l’arabe. Aucun audio d’une personne réelle n’est nécessaire. Les deux environnements isolés évitent de modifier PyTorch du kernel ; le filigrane Chatterbox est conservé. Sources officielles et licences : [Qwen3-TTS](https://github.com/QwenLM/Qwen3-TTS) (Apache 2.0), [Chatterbox](https://github.com/resemble-ai/chatterbox) (MIT). Révisions et provenance enregistrées.

Votre session Colab consomme ses unités habituelles ; aucun achat ni clé API n’est ajouté. L’onglet doit rester connecté pendant la création ; aucune exécution en arrière-plan après arrêt de Colab n’est promise. Les fichiers déjà sauvegardés restent dans Drive. Cette conversation ne reçoit pas automatiquement les fichiers de votre Drive.

Les moteurs ont été essayés dans les notebooks précédents par le propriétaire. Cette nouvelle interface à bouton est vérifiée syntaxiquement et par simulation sans modèles ; connexion Drive, widgets et génération L4 restent à vérifier dans votre session. En cas d’erreur, le diagnostic est affiché et les voix sauvegardées sont conservées.

Les essais restent temporaires dans la session Colab. Après écoute, cliquez sur **Enregistrer cette voix** sous les lecteurs pour conserver la référence, les WAV et les JSON dans Drive. Sans ce clic, rien n’est ajouté à votre bibliothèque Drive ; les essais temporaires peuvent disparaître à la déconnexion. Le panneau reste ouvert pour les créations suivantes.

Après un enregistrement réussi, les lecteurs et résultats des essais disparaissent ; le tableau de création reste disponible avec vos choix. En cas d’erreur, les lecteurs restent accessibles pour réessayer.
